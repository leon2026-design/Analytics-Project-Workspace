### Auto API browser (prints all functions)
Run this cell any time you add new functions. It discovers functions defined in each module and prints their source, so new code shows up automatically.

```python
# Auto-discover and print all functions from our backend modules
import inspect

def print_api_for_module(mod, title=None, include_private=False):
    if title:
        print(f"\n===== {title} =====")
    funcs = [
        (name, obj) for name, obj in inspect.getmembers(mod, inspect.isfunction)
        if obj.__module__ == mod.__name__ and (include_private or not name.startswith("_"))
    ]
    if not funcs:
        print("(no functions found)")
        return
    for name, func in sorted(funcs):
        print(f"\n# {mod.__name__}.{name}\n")
        try:
            print(inspect.getsource(func))
        except OSError:
            print(f"Source not available for {mod.__name__}.{name}")

print_api_for_module(P, "backend.preprocessing")
print_api_for_module(TM, "backend.train_model")
print_api_for_module(PR, "backend.predict")
```
